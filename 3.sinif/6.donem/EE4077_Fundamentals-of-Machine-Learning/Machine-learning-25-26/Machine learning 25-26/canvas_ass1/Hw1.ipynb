```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error

# Load the Car datasets
X_train_car = pd.read_csv('Xtraining_car.csv')
y_train_car = pd.read_csv('Ytraining_car.csv')
X_val_car = pd.read_csv('Xvalidation_car.csv')
y_val_car = pd.read_csv('Yvalidation_car.csv')

# Load the Polynomial Dataset
X_train_poly = pd.read_csv('Xtraining_poly.csv')
y_train_poly = pd.read_csv('Ytraining_poly.csv')
X_val_poly = pd.read_csv('Xvalidation_poly.csv')
y_val_poly = pd.read_csv('Yvalidation_poly.csv')
```

**Question 1:** Ridge Regression Model Fitting

To demonstrate the ridge regression estimation, we use a dataset  that provides gas mileage, horsepower, and other information for 392 vehicles taken from the StatLib library which is maintained at Carnegie Mellon University. The dataset was used in the 1983 American Statistical Association Exposition. The original dataset has 397 observations, of which 5 have missing values for the variable "horsepower". These rows are removed here. The original dataset is available as a CSV file, as well as at https://www.statlearning.com. Each row in the CSV file was formed based on the following 9 variables: 

* mpg: miles per gallon
* cylinders: Number of cylinders between 4 and 8
* displacement: Engine displacement (cu. inches)
* horsepower: Engine horsepower
* weight: Vehicle weight (lbs.)
* acceleration: Time to accelerate from 0 to 60 mph (sec.)
* year: Model year (modulo 100)
* origin: Origin of car (1. American, 2. European, 3. Japanese)
* name: Vehicle name

In this problem, we will implement  the Ridge Regression estimator  with mpg as the response (dependent variable) and cylinders, displacements, horsepower, weight, and acceleration  as the predictors (independent variables). First, we divide the dataset into the training, validation and testing datasets such that the training dataset (files: Ytraining_car.csv, Xtraining_car.csv) consists of 312 observations,   validation set (files: Yvalidation_car.csv, Xvalidation_car.csv) and  testing set (files: Ytesting_car.csv, Xtesting_car.csv) each contains 40 observations.

Implement the fit_ridge function using sklearn.linear_model.Ridge with the following names, inputs, and outputs: $w = \text{fit\_ridge}(X,y,\lambda)$ which calculates the model parameters  given the training dataset $(X,y)$ and a value for the regularization parameter.

```python
def fit_ridge(X, y, lmbda):
    model = Ridge(alpha=lmbda)
    model.fit(X, y)
    return model
```

**Question 2:** Ridge RMSE Evaluation

Calculate the Root Mean Squared Error (RMSE) on the validation set for a given $\lambda$.

```python
def get_rmse(model, X_val, y_val):
    y_pred = model.predict(X_val)
    rmse = np.sqrt(mean_squared_error(y_val, y_pred))
    return rmse
```

**Question 3:** Finding the Optimal Lambda

Iterate through $\lambda$ values from 0.001 to 2.0 with a step of 0.002 and find the $\lambda$ that minimizes validation RMSE.

```python
lambdas = np.arange(0.001, 2.001, 0.002)
rmses = []

for lmbda in lambdas:
    model = fit_ridge(X_train_car, y_train_car, lmbda)
    rmse = get_rmse(model, X_val_car, y_val_car)
    rmses.append(rmse)

best_index = np.argmin(rmses)
best_lambda = lambdas[best_index]
best_rmse = rmses[best_index]
```

**Question 4:** Polynomial Model Pipeline

In the class, we have learned that by using non-linear basis functions, we can fit a non-linear model as if it is a linear model. In this problem, we are given a training data set $D$ that consists of $n$ points, $(x_i, y_i)$ for $i=1,...,n$ and we consider a polynomial regression problem as follows: $$y_{i}=w_{0}+w_{1}x_{i}+...+w_{k}x_{i}^{k}$$ Now, we want to implement polynomial regression on 40 data points. We  divide  the dataset into two parts: training set consisting of 32 points (Files: Xtraining_poly.csv, Ytraining_poly.csv) and validation set  with 8 points (Files: Xvalidation_poly.csv, Yvalidation_poly.csv). 

Implement a function that fits a polynomial model on the training dataset given the polynomial degree $k$, and outputs the weight terms $w$. The function returns a scikit-learn pipeline with PolynomialFeatures and Ridge (set alpha=0.0). 

```python
def fit_polynomial(X, y, k):
    model = make_pipeline(
        PolynomialFeatures(degree=k),
        Ridge(alpha=0.0)
    )
    model.fit(X, y)
    return model
```

**Question 5:** Polynomial RSS Calculation

For a given $k$, obtain the RSS error on the validation set.

```python
def calculate_rss(model, X, y):
    y_pred = model.predict(X)
    rss = np.sum((y.values - y_pred) ** 2)
    return rss
```

**Question 6:** Optimal Poılynomial Degree

Find the degree $k \in \{1, \dots, 10\}$ that minimizes the validation RSS.

```python
ks = range(1, 11)
validation_errors = []

for k in ks:
    model = fit_polynomial(X_train_poly, y_train_poly, k)
    rss = calculate_rss(model, X_val_poly, y_val_poly)
    validation_errors.append(rss)

best_k = list(ks)[np.argmin(validation_errors)]
best_validation_error = validation_errors[np.argmin(validation_errors)]
```
